```python
import autogen
from autogen import ConversableAgent
from IPython.display import display, Markdown
```

```python
from getpass import getpass
import os 

os.environ["OPENAI_API_KEY"] = getpass("Enter OpenAI API Key: ")
```

```python
# Setting GPT's configuration
config_list_gpt = {
    "config_list": [{"model": "gpt-4o-mini", "temperature": 0}],
}
```

```python
# define your first agent
agent_content_creator = ConversableAgent(
    name="Content_Creator",
    system_message="""
                    You are the Content Creator Agent. Your role is to draft conent on topics involving Generative AI.
                    Ensure the content is clear, concise, and technically accurate.
                    """,
    llm_config=config_list_gpt,
    code_execution_config=False,
    human_input_mode="NEVER", 
)
```

```python
critic_agent = ConversableAgent(
    name="Content_Critic_Agent",
    system_message="""
                    You are the Content Critic Agent. Your role is to evalute the content drafted by the Content Creator Agent.
                    Provide feedback on language and technical correctness, and suggest improvements.
                    """,
    llm_config=config_list_gpt,
    code_execution_config=False,
    human_input_mode="NEVER",
)
```

```python
messages = ["What is the role of AGI in Agentic AI."]

# Initiate the conversation
result = agent_content_creator.initiate_chat(critic_agent, 
                               message= messages[0], 
                               max_turns=3,
                               silent=False,
                               summary_method="reflection_with_llm")
```
