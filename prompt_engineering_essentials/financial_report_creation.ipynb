```python
import fitz
from PIL import Image

doc = fitz.open("Financial Image Scans.pdf")
images = []

for page_number in range(len(doc)):
    page = doc[page_number]
    pix = page.get_pixmap(dpi=200)
    img = Image.frombytes("RGB", [pix.width, pix.height], pix.samples)
    images.append(img)

print("Loaded", len(images), "page images")
```

```python
first_image = images[0]
first_image.show()
```

```python
prompt = """
You are a financial analyst tasked with analyzing a set of images containing financial data, including graphs, tables, and text. Your goal is to generate a comprehensive financial report summarizing the key insights from all provided images.

**Analysis Instructions:**
- Examine each image carefully for textual content, numerical data, charts, graphs, and tables.
- Identify key financial metrics, trends, patterns, and anomalies across all images.
- Cross-reference information between images to ensure consistency and derive holistic insights.
- Pay special attention to:
  - Revenue, expenses, profits, and losses
  - Growth rates, percentages, and ratios
  - Time-series data in graphs (e.g., trends over months/quarters/years)
  - Comparative data in tables (e.g., year-over-year comparisons)
  - Any highlighted or emphasized data points

**Report Structure:**
Organize your response in the following clear sections:

1. **Executive Summary**: A concise overview (2-3 sentences) of the overall financial health and key takeaways from the images.

2. **Key Metrics and Figures**: List and explain the most important numerical data extracted from the images, including sources (e.g., "From Image 1, Table A").

3. **Trends and Patterns**: Describe any observed trends, growth patterns, or changes over time based on graphs and comparative data.

4. **Strengths and Opportunities**: Highlight positive aspects, strengths, or growth opportunities identified.

5. **Risks and Concerns**: Note any potential risks, declines, anomalies, or areas of concern.

6. **Recommendations**: Provide actionable insights or suggestions based on the analysis.

**Guidelines:**
- Be objective, data-driven, and professional in tone.
- Use bullet points or numbered lists for clarity where appropriate.
- If data is incomplete or unclear in an image, note it and base insights on available information.
- Ensure the report is coherent and synthesizes information from all images collectively.
- Keep the report concise yet comprehensive (aim for 500-800 words total).

Generate the financial report based on the provided images.
"""
```

```python
response = gemini.generate_content(contents=[prompt] + images)
```

```python
from IPython.display import Markdown, display
display(Markdown(response.text))
```
